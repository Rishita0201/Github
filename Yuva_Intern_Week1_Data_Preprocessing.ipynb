```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values in Titanic Dataset")
plt.xlabel("Columns")
plt.ylabel("Records")

plt.show()
```

```python
missing_values = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

print(missing_report)
```

```python
print("Missing values before cleaning:")
print(df[["Age", "Cabin", "Embarked"]].isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
print("Missing values after cleaning:")
print(df[["Age", "Cabin", "Embarked"]].isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
duplicates = df[df.duplicated()]

print(duplicates)
```

```python
print("Age Statistics:")
print(df["Age"].describe())
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Age"])

plt.title("Box Plot of Age")
plt.xlabel("Age")

plt.show()
```

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Age"] < lower_bound) |
    (df["Age"] > upper_bound)
]

print("Number of Age outliers:", len(outliers))
```

```python
print("Outlier Age values:")
print(outliers["Age"].sort_values().to_list())
```

```python
print("Number of statistical outliers:", len(outliers))
```

```python
print("Minimum outlier age:", outliers["Age"].min())
print("Maximum outlier age:", outliers["Age"].max())
```

```python
print("Outlier records:")
display(outliers[["PassengerId", "Name", "Sex", "Age", "Pclass"]].sort_values("Age"))
```

```python
df["Sex"] = df["Sex"].map({
    "male": 0,
    "female": 1
})
```

```python
print(df["Sex"].value_counts())
```

```python
df = pd.get_dummies(
    df,
    columns=["Embarked"],
    drop_first=True
)
```

```python
analysis_df = df.drop(
    columns=["PassengerId", "Name", "Ticket", "Cabin"]
)

print("Analysis-ready dataset shape:", analysis_df.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

analysis_df[["Age", "Fare"]] = scaler.fit_transform(
    analysis_df[["Age", "Fare"]]
)

print("Age and Fare standardized successfully.")
```
